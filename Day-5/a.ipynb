#2.Problem Statement: Write a Python program to create a Vehicle class with two instance attributes: max_speed and mileage. Create an object of the class and print both attributes.

```python
class Vehicle:

    def __init__(self, name, max_speed, mileage):
        self.name = name
        self.max_speed = max_speed
        self.mileage = mileage

vehicle1 = Vehicle("Tesla model S", 250, 19)
print(vehicle1.mileage)
print(vehicle1.max_speed)
```

#3.Problem Statement:  Write a Python program to create a Rectangle class with length and width as instance attributes, and two methods: area() that returns the area and perimeter() that returns the perimeter.

```python
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width


    def perimeter(self):
        return 2 * (self.length + self.width)


rect1 = Rectangle(4, 5)
rect2 = Rectangle(6, 7)
print(rect1.area())
print(rect1.perimeter())
print(rect2.area())
print(rect2.perimeter())
```

#4.Problem Statement: Write a Python program to create a Student class that stores a student’s name and a list of marks. Add a method average() that calculates and returns the average of all marks.

```python
class Student:
    def __init__(self,name:str,marks:list):
        self.name = name
        self.marks = marks

    def average(self):
        return sum(self.marks) / len(self.marks)


student1 = Student("Aryan",[76, 67, 89, 97, 50])
print(student1.average())
```

#5. Problem Statement : Write a Python program to create a Product class with three instance attributes: name, price, and quantity. Add a method total_value() that returns the total stock value by multiplying price by quantity.


```python
class Product:

    def __init__(self, name, price, quantity):
        self.name = name
        self.price = price
        self.quantity = quantity


    def total_value(self):
        return abs(self.price * self.quantity)

product = Product("Ipad 10th Gen", 310800, 5)
print(product.total_value())
```
